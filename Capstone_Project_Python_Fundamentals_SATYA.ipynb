```python
#TASK 2: The cost column in the dataframe “Project” has some missing values. Your task is to compute these missing values. Replace the missing values by running average. You should use the “For” loop for this task. 

import pandas as pd  # Import the Pandas library

df_p = pd.read_csv('Project.csv')  # Load  Project dataset CSV into active workspace memory

running_sum = 0  # Initialize variable tracking the sum of valid costs found up to current index
running_count = 0  # Initialize variable tracking the count of valid rows found up to current index

for i in range(len(df_p)):  # Loop step-by-step through every row index position across the dataset
    current_cost = df_p.loc[i, 'Cost']  # Extract the individual project cost entry value for the current active row
    
    if not pd.isna(current_cost):  # Check if the extracted cost cell value is a valid numeric number
        running_sum += current_cost  # Add the current valid project cost amount to our running sum tracker
        running_count += 1  # Increment our row counter tracker of valid records by one unit
    else:  # Run alternative mathematical logic if the current cost cell matches missing NaN criteria
        running_avg = running_sum / running_count  # Calculate running average by dividing total sum by total valid count
        df_p.loc[i, 'Cost'] = running_avg  # Insert the calculated running average back into the empty dataframe cell
        running_sum += running_avg  # Add the newly calculated average value directly into our rolling sum memory
        running_count += 1  # Increment our running row counter to properly include this newly updated cell

# Force notebook to show clean float currencies instead of raw scientific notation
pd.set_option('display.float_format', lambda x: '%.2f' % x)

df_p.to_csv('Project.csv', index=False)  # Save the finalized project dataframe back to its CSV layer
df_p  # Print out  project table
```

```python
#TASK 3: Split the name column in the Employee dataframe into two new columns “First Name”, and “Last Name” and remove the older “name” column. 

import pandas as pd # Import the Pandas library

df_emp = pd.read_csv("Employee.csv") # Load dataset from Task 1

df_emp[["First Name", "Last Name"]] = df_emp["Name"].str.split(" ", n=1, expand=True) # Split at the first space into the two columns

df_emp.drop(columns=["Name"], inplace=True) # Drop old column and save file

df_emp.to_csv("Employee.csv", index=False) # Saves the updated DataFrame to a CSV file while omitting row index numbers

df_emp #Evaluates and displays the DataFrame

```

```python
#TASK 4: Join all three dataframes in one single dataframe. Name it “Final” 

import pandas as pd  # Import the Pandas library 

df_p = pd.read_csv('Project.csv')  # Load  Project dataset CSV into active memory
df_e = pd.read_csv('Employee.csv')  # Load  split-name Employee dataset CSV into active memory
df_s = pd.read_csv('Seniority.csv')  # Load  raw Seniority level dataset CSV into active memory

df_merged_emp = pd.merge(df_e, df_s, on='ID')  # Join the edited employee dataframe with the seniority dataframe matching along their unique shared employee ID field
Final = pd.merge(df_merged_emp, df_p, on='ID')  # Merge that intermediate table together with our updated project data table matching along the same ID column fields

Final.to_csv('Final.csv', index=False)  # Save the consolidated source-of-truth table directly as a new CSV layer
pd.set_option('display.float_format', lambda x: '%.2f' % x)  # Configure display format options to get clean decimals instead of unreadable scientific notations

Final  # Print out all 14 rows of table
```

```python
#TASK 5: Add a new bonus column in the Final dataframe. Give a 5% bonus concerning project cost only to employees who have finished the projects.

import numpy as np  # Import the NumPy library 
import pandas as pd  # Import the Pandas library 

Final = pd.read_csv('Final.csv')  # Load combined master dataset from the saved CSV file

# Compute the bonus: 5% of cost if Status is Finished, otherwise set it strictly to 0.0
Final['Bonus'] = np.where(Final['Status'] == 'Finished', Final['Cost'] * 0.05, 0.0)

Final.to_csv('Final.csv', index=False)  # Save updated dataframe with the new bonus column back to CSV
Final  # Print out all rows of table
```

```python
#TASK 6: Demote the designation level by 1, whose projects have status “fail”. Delete the employees record whose designation level is above 4.

import numpy as np  # Import the NumPy library 
import pandas as pd  # Import the Pandas library 

Final = pd.read_csv('Final.csv')  # Load your combined master dataset from the saved CSV file

# If status is Failed, add 1 to Designation Level; otherwise, keep the current Designation Level intact
Final['Designation Level'] = np.where(Final['Status'] == 'Failed', Final['Designation Level'] + 1, Final['Designation Level'])

# Drop rows by preserving only those records where the Designation Level does not exceed 4
Final = Final[Final['Designation Level'] <= 4]

Final.to_csv('Final.csv', index=False)  # Save penalty dataframe back to the CSV layer
Final  
```

```python
#TASK 7: Add “Mr.” and “Mrs.” to the first name column and drop the gender column.

import numpy as np  # Import the NumPy library 
import pandas as pd  # Import the Pandas library

Final = pd.read_csv('Final.csv')  # Load dataset from the saved CSV file

if 'Gender' in Final.columns:  # Check if the Gender column exists before transforming to prevent re-run errors
    Final['First Name'] = np.where(Final['Gender'] == 'M', 'Mr. ' + Final['First Name'], 'Mrs. ' + Final['First Name'])  # Prepend 'Mr. ' or 'Mrs. ' to the First Name based on Gender criteria
    Final = Final.drop(columns=['Gender'], errors='ignore')  # Drop binary Gender classification column directly from  memory to clean the layout

Final.to_csv('Final.csv', index=False)  # Save  dataframe structure back to the CSV layer
Final  # Print out all rows of table 
```

```python
#TASK 8: Promote designation level by 1 for the employees whose age is more than 29 years using IF condition. 

import numpy as np  # Import the NumPy library 
import pandas as pd  # Import the Pandas library 

Final = pd.read_csv('Final.csv')  # Load dataset from the saved CSV file

# Promote employees older than 29 by subtracting 1 from their Designation Level; otherwise, keep it the same
Final['Designation Level'] = np.where(Final['Age'] > 29, Final['Designation Level'] - 1, Final['Designation Level'])

Final.to_csv('Final.csv', index=False)  # Save promotion modified master dataframe back to CSV
Final  # Print 
```

```python
#TASK 9: Add the cost of all projects for each Employee and save it in new dataframe “TotalProjCost” with three columns ID, First Name, and Total cost

import pandas as pd  # Import the Pandas library

Final = pd.read_csv('Final.csv')  # Load  promotion modified master dataset from the saved CSV file

# Group by ID and First Name, calculate the sum of the Cost column, and reset the index to keep a flat table layout
summary_df = Final.groupby(['ID', 'First Name'])['Cost'].sum().reset_index()

# Rename the aggregated sum calculation column directly to 'Total cost' to match submission criteria
summary_df = summary_df.rename(columns={'Cost': 'Total cost'})

summary_df.to_csv('Summary_Task9.csv', index=False)  # Save the  aggregate summary layer directly to a fresh CSV file
summary_df  # Display the complete generated summary dataset output metrics on notebook screen
```
